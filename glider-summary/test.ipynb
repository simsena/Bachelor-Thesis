```python
import xarray as xr
import matplotlib.pyplot as plt

ds = xr.open_dataset("/Users/sensen/Desktop/BA/glider-summary/raw_data/ifm02_depl17_glider_gridded.nc")
ds

```

```python
import matplotlib.pyplot as plt, cmocean, numpy as np
fig, axs = plt.subplots(2, 1, figsize=(8, 2))
grad = np.linspace(0, 1, 256).reshape(1, -1)
axs[0].imshow(grad, aspect="auto", cmap="jet");     axs[0].set_title("jet");     axs[0].set_axis_off()
axs[1].imshow(grad, aspect="auto", cmap=cmocean.cm.thermal); axs[1].set_title("cmocean thermal"); axs[1].set_axis_off()
plt.savefig("cmap_compare.png", dpi=150, bbox_inches="tight")
```

```python
import cmocean

var = "TEMP"   # oder "FLU2", "PSAL", ...

fig, axs = plt.subplots(1, 2, figsize=(11, 4.2), constrained_layout=True)

ds[var].plot(ax=axs[0], y="PRES", cmap="jet",
             yincrease=False, robust=True,
             cbar_kwargs={"label": "Temperature [°C]"})
axs[0].set_title("jet")

ds[var].plot(ax=axs[1], y="PRES", cmap=cmocean.cm.thermal,
             yincrease=False, robust=True,
             cbar_kwargs={"label": "Temperature [°C]"})
axs[1].set_title("cmocean thermal")

fig.savefig("cmap_compare.png", dpi=150, bbox_inches="tight")
plt.show()
```
